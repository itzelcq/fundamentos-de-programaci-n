El programa crea una matriz de multiplicación del 1 al 10, mostrando los resultados de cada combinación de filas y columnas. Después, solicita al usuario dos números y utiliza la matriz para obtener y mostrar el resultado de su multiplicación.

```python
#Tabla de pitagoras

matriz = []

for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        fila.append(i * j)
    matriz.append(fila)

def encabezado():

    print("    ", end="") 
    for c in range(1, 11): # Columnas
        print(f"{c:4d}", end="") 
    
    print()
    
    for f in range(10): # Filas
        print(f"{f + 1:3d}|", end="")
        
        for c in range(10):
            print(f"{matriz[f][c]:4d}", end="") 
        
        print()


def multiplicar(matriz, n1, n2):
    return matriz[n1 - 1][n2 - 1]

encabezado()

n1 = int(input("Ingrese un numero:" ))
n2 = int(input("Ingrese otro numero: "))

resultado = multiplicar(matriz, n1, n2)

print({resultado})
```
