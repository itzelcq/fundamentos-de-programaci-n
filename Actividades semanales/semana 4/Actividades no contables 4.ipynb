````python

#preguntar cuantas calificaciones desea promedia , y promediarlas

def promedio(sumatoria, n):
    return sumatoria / n

n=int(input("¿Cuantas calificaciones quieres promediar?"))
sumatoria=0

print(f"se van a pedir {n} calificaciones")

for c in range (1, n+1):
   calific=float(input(f"dame la calificacion {c}"))
   sumatoria=sumatoria + calific

print(f"el promedio de {n} calificaciones es {promedio(sumatoria, n)}")
````

````python
filas=int(input("¿Cuantas filas quieres en la matriz?"))
columnas= 5

matriz=[]
for f in range(filas):
    fila_actual=[]
    for c in range(columnas):
        valor=int(input(f"dame el valor de la posicion [{f}][{c}]"))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print ("Impresion de la matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print("")
````

````python
dimension=int(input("¿Cuantas filas quieres en la matriz?"))
columnas= 5

matriz=[]
for f in range(filas):
    fila_actual=[]
    for c in range(columnas):
        valor=0
        if f==c:
            valor=1
        fila_actual.append(valor)
    matriz.append(fila_actual)

print ("Impresion de la matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print("")
````
