Uso de tuplas
1. Crear una tupla llamada `numeros` con un mínimo de **cinco elementos numéricos**
de partida.
2. Acceder e imprimir en pantalla el **tercer elemento** de la tupla.
3. Capturar **dos números adicionales** ingresados por el usuario mediante
`input()` y anexarlos para crear una **nueva tupla**.
4. Convertir la tupla en **lista**, aplicar un **ordenamiento** a sus elementos y
mostrarla.
5. Diseñar una **función** que tome la tupla de números como parámetro, realice la
**suma** de todos sus elementos y **retorne** dicho valor para mostrarlo en
pantalla.

```python
numeros = (10, 20, 30, 40, 50)

print("tercer numero:", numeros[2])

numero1 = int(input("ingrese un numero: "))
numero2 = int(input("ingrese el segundo: "))

numeros = numeros + (numero1, numero2)

lista = list(numeros)
lista.sort()

print("numeros ordenados:", lista)

def suma_tupla(numeros):
    return sum(numeros)

resultado = suma_tupla(numeros)

print("la suma de todos los elementos, es:", resultado)
```

Uso de diccionarios
1. Crear un diccionario llamado `contactos` con al menos **tres registros
iniciales** (clave: nombre del contacto, valor: número de teléfono).
2. Permitir la **captura y adición de un nuevo contacto** al diccionario mediante
la consola.
3. **Iterar sobre las claves** del diccionario e imprimir en pantalla
exclusivamente los **nombres** de los contactos registrados.

```python
contactos = {
    "vania": "442-980-0989",
    "nico": "442-364-9229",
    "aldo": "442-274-9239"
}

nombre = input("Ingrese el nombre del nuevo contacto: ")
telefono = input("Ingresa el numero de telefono: ")
contactos[nombre] = telefono

print("\nCONTACTOS:")

for contacto in contactos:
    print(contacto)

def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "el contacto no es valido")

nombre_buscar = input("\nIngrese el nombre del contacto que desee buscar: ")

resultado = buscar_contacto(contactos, nombre_buscar)

print("numero de telefono:", resultado)
```

```python
numero1 = input("ingrese el primer número: ")
print (numero1)


numero2 = input("ingrese el segundo número: ")
print (numero2)

try:
    if not numero1.isdigit() or not numero2.isdigit():
        print("NUMERO INVALIDO!!, SOLO NUMEROS ENTEROS.")
    
    elif numero2 == "0":
        print("NUMERO INVALIDO, NO PUEDE SER CERO")
    else:
        num1 = int(numero1)
        num2 = int(numero2)
        print("la suma de ambos numeross es:", num1 + num2)

except Exception:
    print("OCURRIO UN ERROR =( ")
```

2.4 Uso de strings
1. Crear una variable de texto llamada `mensaje`.
2. Imprimir en pantalla la **longitud** del string utilizando la función `len()`.
3. Utilizar un método de strings para transformar la totalidad del mensaje a
**mayúsculas**.
4. Utilizar un método de strings para **buscar y reemplazar** una palabra clave del
mensaje por otra palabra de tu elección.
5. Diseñar una **función** que reciba un string como parámetro y retorne de forma
correcta la **cantidad de palabras** que contiene, mostrando el resultado final.

```python
mensaje = "me voy a dar de baja de la carrera pipipi"

print("longitud del mensaje:", len(mensaje))

mensaje = mensaje.upper()
print("Mayusculas:", mensaje)

mensaje1 = mensaje.replace("PIPIPI", "voy a llorar")
print("modificacion:", mensaje1) 

def contar_palabras(texto):
    return len(texto.split())

cantidad = contar_palabras(mensaje)
print("cantidad de palabras:", cantidad)
```
