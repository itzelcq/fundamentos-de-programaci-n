Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

```python
votos = {
    "rojo": 0,
    "azul": 0 ,
    "verde": 0,
}

for i in range (5): 
    color = input ("Ingrese color del voto: ")
    if color in votos:
        votos [color] += 1
    else: 
        print ("COLOR INVALIDO!!")

ganador = ()
votos2 = -1

for color, total in votos.items():
    if total > votos2:
        votos2 = total 
        ganador = color 

print (f"el ganador fue {ganador}!! y tiene {votos2} votos!!")
```

Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque
`try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje
controlado y repite la solicitud con un ciclo `while`. Luego pide la división del
primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
while True:
    try:
        numero1 = int(input("ingrese el primer numero: "))
        numero2 = int(input("ingrese el segundo numero: "))

        print (numero1)
        print (numero2)

        resultado = numero1 / numero2

        print(resultado)
        break

    except ValueError:
        print("INVALIDO!!! ingrese un numero entero \n")

    except ZeroDivisionError:
        print("INVALIDO!!! no se puede divir entre cero\n")
```

Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con `len()`, la
convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre
cuántas palabras contiene usando una función que retorne el conteo.

```python
def palabra(texto):
    return len(texto.split())

frase = input("ingrese una frase, pon hola: ")

print(f"la longitud de la frase es: {len(frase)}")

print(f"la frase en mayusculas es: {frase.upper()}")

reemplazo = frase.replace("hola", "aloha")
print(f"el texto reemplazado es: {reemplazo}")

total = palabra(frase)
print(f"el total de las palabras: {total}")
```
