Extra 1: Sistema de calificaciones con tuplas

Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.
**Entrada:**
```
Nueva calificación 1: 8.5
Nueva calificación 2: 9.5

````python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)       
print(calificaciones[2]) 

calif1 = float(input("Ingrese una calificacion: "))   
calif2 = float(input("Ingrese otra calificacion: "))

nueva_tupla = calificaciones + (calif1, calif2)     
print( nueva_tupla)               
lista_calificaciones = list(nueva_tupla)           

lista_calificaciones.sort(reverse=True)   
print("-----------------------------------------------------")            
print("Lista ordenada :", lista_calificaciones) 
print("-----------------------------------------------------")    

def sumar_calificaciones(lista):                      
    return sum(lista)

suma_total = sumar_calificaciones(lista_calificaciones)   




````

Extra 2: Agenda de contactos con búsqueda
**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto

````python
agenda = {                                 
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nombre = input("Nombre del nuevo contacto: ")  
telefono = input("Teléfono del nuevo contacto: ")  

agenda[nombre] = telefono             
nombres = agenda.keys()                      

print("Contactos registrados:", ", ".join(nombres))

def buscar_telefono(nombre):             
    return agenda.get(nombre)          

nombre1 = input("Nombre a buscar: ")         
telefono = buscar_telefono(nombre1)
print(f"El teléfono de {nombre1} es: {telefono}")   
````

Extra 3: Calculadora segura con manejo de excepciones
**Enunciado:** Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.

````python
try:                               
    numero1 = int(input("Primer número entero: "))  
    numero2 = int(input("Segundo número entero: "))

    resultado = numero1 / numero2      
    print(f"{numero1} dividido entre {numero2} es: {resultado}")

except ValueError:                              
    print("Error: debes ingresar números enteros.")

except ZeroDivisionError:                  
    print("Error: no se puede dividir entre 0 entre cero.")
    
````

Extra 4: Analizador de mensajes con strings
**Enunciado:** Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.


````python
mensaje = "Python es un lenguaje poderoso"     
print("Longitud del mensaje:", len(mensaje))    
mensaje_mayusculas = mensaje.upper()            
print("En mayúsculas:", mensaje_mayusculas)

texto_reemplazado = mensaje.replace("Python", "programación")  
print("Texto reemplazado:", texto_reemplazado)  

def contar_palabras(texto):      
    return len(texto.split())

total_palabras = contar_palabras(mensaje)         
print("Palabras totales:", total_palabras)
````
