Ejercicio 11: Suma con centinela 0
**Enunciado:** Crea un programa que sume números positivos ingresados por el
usuario y se detenga cuando ingrese `0`, mostrando la suma total

```python
suma=0
while True:
    num = float(input("Ingrese un valor(Ingresa 0 para terminar)"))
    if num == 0:
       break
    suma += 0

print(f"el total de la suma es {suma}")
```
