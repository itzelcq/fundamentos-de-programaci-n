Extra 1: Control de aforo con break y continue (while)
**Enunciado:** Crea un programa que procese visitantes del museo con un ciclo
`while`. Por cada visitante se captura el costo de su boleto ($0 para menores de 3
años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis,
muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa
`continue`). Si el total acumulado alcanza o supera $100, detén el ciclo con
`break` y muestra el total.

````python
total = 0   
contador = 1

while True:
    boleto = float(input (f"precio del boleto {contador}:"))
    if boleto == 0:
        print("menor de 3 años, sin cargo")
        contador += 1
        continue

    total += boleto 
    contador += 1
    if total >= 100:
        print ("aforo alcanzado, se detiene el registro")    
        break 

print (f"total acumulado: ${total:.2f}")
````

Extra 2: Estadística de visitantes con for
**Enunciado:** Crea un programa que pida el número total de visitantes y, con un
ciclo `for`, capture la edad de cada uno. Debe contar cuántos son adultos (18 años
o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos
resultados con dos decimales.

````python
visitantes = int(input("Ingrese el número de visitantes: "))
adultos=0
suma_edades = 0
for i in range(1, visitantes + 1):
    edad = int(input(f"Ingrese la edad del visitante {i}: "))
    suma_edades += edad
    if edad >= 18:
        adultos+= 1
promedio = suma_edades / visitantes

print(f"Adultos: {adultos}")
print(f"Promedio de edades: {promedio:.2f}")
````
