```python
visitantes = int(input ("agrega el total de visitantes: "))
total = 0
contador=1

for i in range (1,visitantes+1):
     print(f"\n Visitante:{i}")
     if visitantes > 5:
         print("DEMASIADAS PERSONAS")
         break
     edad = int(input(f"ingrese la edad del visitante {i}: "))
   
     if edad <=3:
        costo = 0
     elif edad <=18:
        costo = 30
     else :
        costo = 45
     if costo == 0:
            print("menor de 3 años, ENTRA GRATIS!!")
            contador += 1
            continue
   

     tipo = input(f"ingrese el tipo de visitante (estudiante, profesor,adulto_mayor) {i}: ")
     descuento=0
     if tipo == "estudiante" or tipo == "profesor":
            descuento = costo * 0.10
     elif tipo == "adulto_mayor":
            descuento = costo * 0.12
        
     precioporpersona = costo - descuento


     print(f"precio por persona con descuento: $ {precioporpersona}")
     total += precioporpersona

print("------------------------------------------------")

print(f"TOTAL A PAGAR POR TODO EL GRUPO: ${total}")

print("------------------------------------------------")
```
