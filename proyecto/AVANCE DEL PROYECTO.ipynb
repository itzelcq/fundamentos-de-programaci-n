2.1 Problemática

Talent for Talent ha identificado que muchas empresas invierten en reclutamiento, y capacitacion, pero aun así continúan perdiendo trabajadores importantes.

La propuesta consiste en desarrollar un programa sencillo que permita calcular un nivel de riesgo de renuncia para cada trabajador, utilizando diferentes variables y un sistema de puntuación.

2.2 Objetivo general
Desarrollar un prototipo en Python que permita identificar el nivel de riesgo de rotación de un trabajador mediante una puntuación basada en diferentes factores relacionados con su experiencia laboral.

11. Los colaboradores con riesgo alto deberán recibir una intervención prioritaria.
