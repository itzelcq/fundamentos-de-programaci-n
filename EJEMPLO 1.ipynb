```python
Total_de_cuentas=float(input("Ingrese total de la cuenta:"))
Propina_a_dejar=float(input("Ingrese el porcentaje de propina que desea dejar:"))
numero_de_peronas=float(input("Ingrese el total de personas que pagaran la cuenta:"))

propina= Total_de_cuentas * Propina_a_dejar /100
total_de_la_cuenta_con_propina= Total_de_cuentas *(1+Propina_a_dejar)/100
total_a_pagar_por_persona= total_de_la_cuenta_con_propina/numero_de_peronas

print("propina a dejar:",propina)
print("Total de la cuenta con propina:",total_de_la_cuenta_con_propina)
print("Total a pagar por persona:",total_a_pagar_por_persona)
```
